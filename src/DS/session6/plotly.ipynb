```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("train.csv")
df.head(4)
```

```python
fig = px.histogram(df, x='Survived' , color='Sex', title='<b> survived based on gender </b>' , barmode="group")
fig.update_layout(width =700 , height = 500 , bargap = 0.2)
```

```python
fig = px.box(df , y = "Age" , points='all' , title="titanic age box with outliers"  )
fig.show()
```

```python
fig1 = px.box(df , y = "Fare" , points='all' , title="titanic fare box with outliers"  )
fig1.show()
```

```python
col = ['Age']
Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]
normal =df[(df[col] >= Q1 - 1.5*IQR) & (df[col] <= Q3 + 1.5*IQR)]

fig = go.Figure()
fig.add_trace(go.Scatter(y = normal[col] , mode='markers' , name= 'Normal'))

fig.add_trace(go.Scatter(y=outliers , mode="markers" , marker=dict(color = 'red' , size = 10)))
```
