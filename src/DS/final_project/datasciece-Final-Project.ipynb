###  To fix this issue You will fill these values with median for numeric and mode for categorical

```python
df.head(2)
```

```python
# trying to fill missing values in my datset using for loop 

for  col in df.select_dtypes('number').columns:
    if df[col].isnull().any() == True :
        df[col] = df[col].fillna(df[col].median())



cols = ['start_time' ,
        'end_time',
        'start_station_name',
        'end_station_name',
        'user_type',
        'member_gender',
        'bike_share_for_all_trip']

df[cols] = df[cols].astype("category")
check_types(df)


meadian_1 = df[cols].mode()

meadian_1
df[cols].fillna(meadian_1 , inplace=True)



```

```python
#check the Nulls 
check_null(df)
```

# <a id="8"> 8) . What the average duration of trip for the categories of user type?</a>

```python
# Distribution of Age by user types
from preporessing import check_null

#df['duration_sec'] = df['duration_sec'] / 60

avg_duration = df.groupby('user_type')['duration_sec'].mean().round(2)
avg_duration




plt.figure(figsize=(5,3))
sns.barplot(
    x=avg_duration.index,
    y=avg_duration.values
)
plt.title("Average Trip Duration by User Type (Hours)")
plt.xlabel("User Type")
plt.ylabel("Average Duration (Hours)")
plt.show()
```

# <a id="9">    9 ) . How long is the duration of the trip for each gender in minutes?  </a>

```python
# Make visualization to easy get insight

# ------------------------Duration distribution by member gender


df['duration_min'] = df['duration_sec'] * 60  

plt.figure(figsize=(7,4))
sns.histplot(
    data=df,
    x='duration_min',
    hue='member_gender',
    bins=50,
    multiple='stack'
)
plt.title("Distribution of Trip Duration by Gender (Minutes)")
plt.xlabel("Trip Duration (Minutes)")
plt.ylabel("Number of Trips")
plt.show()
```
